# Productionization of ML Systems - Final Exam, Q1 and Q2

## Q1. Model Deployment (10 marks)

Deployment is how a trained model gets its predictions to the people or systems that need them. The main options are:

**Batch / offline prediction.** The model runs on a schedule over many records and saves the results to a table or file, e.g. a nightly churn score for every customer. Use it when nobody needs the answer straight away. It's the simplest and cheapest option.

**Real-time REST API.** The model sits behind an endpoint (FastAPI or Flask). A client sends one record and gets a prediction back in milliseconds, like Q3. Use it when the answer is needed immediately, e.g. scoring a loan application while the customer waits.

**Cloud managed services.** AWS SageMaker, Azure ML, Google Vertex AI or Databricks Model Serving host the model for you and handle servers, scaling and versioning. Use it when you don't want to manage infrastructure. It costs more but saves engineering time.

**Containers.** Docker packages the model, code and libraries so it runs the same everywhere. Kubernetes runs many copies and scales them up and down. Use it when you need reproducibility, or Kubernetes when traffic is high.

**Serverless.** AWS Lambda or Cloud Functions run the model only when a request comes in, and you pay per call. Use it for light or occasional traffic. The downside is a slow first request (cold start) and limits on model size.

**Edge / on-device.** The model runs on a phone, IoT device or browser using TF Lite or ONNX. Use it when there's no internet, very low latency is needed, or data must stay on the device for privacy, e.g. face unlock.

**Embedded in apps or spreadsheets.** The model is called from a tool people already use, like Excel calling an API in Q3. Use it when the users are business staff, since they'll actually use a model that lives in their spreadsheet.

**Choosing.** It mainly comes down to latency, cost and scale. If the answer can wait, batch is cheapest. If it's needed now, use an API, and move to managed services or Kubernetes as traffic grows. Start simple and add complexity only when needed.

## Q2. Model Monitoring and Logging (10 marks)

A model that works well on launch day will get worse over time, and it does this quietly. It doesn't crash, it just gives worse answers. Monitoring and logging are how we notice.

**Why models degrade.** *Data drift* is when the inputs change, e.g. a bank starts getting applicants with much higher incomes than in the training data. *Concept drift* is when the relationship itself changes, e.g. the same debt-to-income ratio becomes riskier during a recession. Data drift can be spotted by comparing new inputs to the training data. Concept drift only shows up once the real outcomes come in.

**What to track.** Model performance (accuracy, AUC) as true outcomes arrive, and the spread of predictions. Also service health: latency, error rates and timeouts. Data quality checks (missing values, out-of-range values, renamed columns) catch a lot of real problems.

**Logging.** Every prediction should be logged with its inputs, output, time and model version. This helps with debugging ("why did this customer get this score?"), and with auditing and compliance. That matters a lot in banking, where a bank may need to explain why a loan was refused. The model version shows whether a drop in performance started when a new model was released.

**Alerts and retraining.** Set thresholds that send an alert, e.g. AUC below 0.75 or error rate above 1%. Retraining can be on a schedule (e.g. monthly) or triggered when drift or performance crosses a threshold. A new model should be tested on a small share of traffic first, with a way to roll back.

**Tools.** Prometheus and Grafana for service metrics and dashboards, MLflow for tracking model versions, Evidently AI for drift reports, and the cloud providers' own monitoring (e.g. SageMaker Model Monitor).

**Example.** A bank trains a loan default model before a recession. When the recession hits, people who look the same on paper are now more likely to default, but the model keeps giving them low risk scores. The API stays up and nothing looks broken, so without monitoring the bank keeps approving risky loans for months. With drift checks and alerts, the change would have been spotted early and the model retrained.
